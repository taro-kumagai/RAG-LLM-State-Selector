```python
%%writefile BehaviorGenerator/behavior_profles.cs
using System;
using System.IO;
using System.Collections.Generic;
using System.Text.Json;

public class BehaviorProfile
{
    public string Name { get; set; }
    public string Description { get; set; }
    public List<string> ScenarioChunks { get; set; }

    public BehaviorProfile(string name, string description)
    {
        Name = name;
        Description = description;
        ScenarioChunks = new List<string>();
    }

    public string ToJson()
    {
        return JsonSerializer.Serialize(this, new JsonSerializerOptions { WriteIndented = true });
    }
}

public class Program
{
    public static void Main()
    {
        var behaviorProfiles = new List<BehaviorProfile>();

        // Building Aggressive_Melee Behavior Profile
        string aggressive_melee_description = "An aggressive agent that prefers to attack rather than defend itself or others. The agent is only capable in melee combat and will not perform any long range attacks";
        var Aggressive_Melee = new BehaviorProfile("Aggressive Melee", aggressive_melee_description);

        Aggressive_Melee.ScenarioChunks.Add("When enemy is low HP, the agent will prioritize that target.");
        Aggressive_Melee.ScenarioChunks.Add("Agent does not attack allies.");
        Aggressive_Melee.ScenarioChunks.Add("Agent tend to target the closest enemy");
        Aggressive_Melee.ScenarioChunks.Add("When there are no enemies, agent will return to starting location");
        Aggressive_Melee.ScenarioChunks.Add("If Enemies are out of range the agent will need to approach its target");

        behaviorProfiles.Add(Aggressive_Melee);

        // Building Doctor Behavior Profile
        string doctor_description = "An agent without any combat capabilities. Helps allies by healing and supporting them.";
        var Doctor = new BehaviorProfile("Doctor", doctor_description);

        Doctor.ScenarioChunks.Add("When an ally has low HP, the agent will prioritize healing that ally.");
        Doctor.ScenarioChunks.Add("When multiple allies are injured, the agent will prioritize the ally with the lowest HP percentage.");
        Doctor.ScenarioChunks.Add("Agent cannot attack enemies under any circumstances.");
        Doctor.ScenarioChunks.Add("When an enemy approaches, the agent will attempt to retreat to a safe location.");
        Doctor.ScenarioChunks.Add("When no allies require healing, the agent will remain near friendly units.");
        Doctor.ScenarioChunks.Add("If an injured ally is out of healing range, the agent will approach that ally.");
        Doctor.ScenarioChunks.Add("Agent will prioritize its own survival when threatened by nearby enemies.");

        behaviorProfiles.Add(Doctor);

        // Building Guardian Behavior Profile
        string guardian_description = "An agent with high defensive capabilties. Protects allies and prioritizes survival. Capable of attacking but will prioritize survival and protection over attacking enemies.";
        var Guardian = new BehaviorProfile("Guardian", guardian_description);

        Guardian.ScenarioChunks.Add("When an ally is threatened by an enemy, the agent will prioritize protecting that ally.");
        Guardian.ScenarioChunks.Add("When multiple allies are threatened, the agent will prioritize protecting the ally with the lowest HP.");
        Guardian.ScenarioChunks.Add("Agent will prioritize defensive actions over offensive actions.");
        Guardian.ScenarioChunks.Add("When an enemy approaches a vulnerable ally, the agent will attempt to intercept that enemy.");
        Guardian.ScenarioChunks.Add("Agent will attack enemies when doing so does not compromise the safety of its allies.");
        Guardian.ScenarioChunks.Add("When the agent has low HP, it will prioritize defensive actions and survival.");
        Guardian.ScenarioChunks.Add("When no allies are threatened, the agent will remain near friendly units.");
        Guardian.ScenarioChunks.Add("Agent will avoid pursuing distant enemies if doing so leaves allies unprotected.");

        behaviorProfiles.Add(Guardian);

        // Converting to json
        foreach (var profile in behaviorProfiles)
        {
            // Converting the profile list to JSON
            string json = JsonSerializer.Serialize(profile, new JsonSerializerOptions { WriteIndented = true });

            // Saving JSON to a file
            File.WriteAllText("behavior_profile_" + profile.Name + ".json", json);
        }

    }
}
```

```python
%%writefile GameSimulator/Character.cs
using System;
using System.IO;
using System.Collections.Generic;

public class CharacterContext
{
    public string Name { get; private set; } = "INVALID_CHARACTER";
    public List<string> Context { get; private set; } = new List<string>();

    public CharacterContext(string name)
    {
        Name = name;
    }
}

public class Character
{
    public string Name { get; private set; } = "INVALID_CHARACTER";
    public State? CurrentState;
    public float HP;
    public float MAXHP;
    public float MeleeAttackRange;
    public float ProjectileAttackRange;
    public float Distance;
    public Character? Target = null;

    public Character(string name)
    {
        Name = name;
    }

    public CharacterContext ConvertToContext()
    {
        var context = new CharacterContext(Name);

        // State
        if (CurrentState != null)
        context.Context.Add("State: " + CurrentState.Name);

        // HP
        if (MAXHP > 0)
            context.Context.Add("HP: " + HP.ToString() + "/" + MAXHP.ToString() + " (" +(HP / MAXHP * 100f).ToString("F1") + "%)");
        else
            context.Context.Add("HP: INVALID HP");

        // MeleeAttackRange
        context.Context.Add("Melee Attack Range: " + MeleeAttackRange.ToString());

        //ProjectileAttackRange
        context.Context.Add("Projectile Attack Range: " + ProjectileAttackRange.ToString());


        // Distance
        context.Context.Add("Distance from Agent: " + Distance.ToString() + "m");

        // Target
        if (Target != null)
            context.Context.Add("Targeting: " + Target.Name);

        return context;
    }
}
```

```python
%%writefile GameSimulator/GameContext.cs
using System;
using System.IO;
using System.Collections.Generic;
using System.Text.Json;


public class GameContext
{
    // List of Valid States for the Agent to transition to
    public List<State> ValidTransitionStates { get; set; } = new List<State>();

    // Agent the is the perspective of the given Game Context
    public CharacterContext? Agent { get; set; } = null;

    // List of Characters in the Game relevant to current situation
    public List<CharacterContext> Characters { get; private set; } = new List<CharacterContext>();
    public void AddCharacter(CharacterContext characterContext) { Characters.Add(characterContext); }

    public string ToJson()
    {
        return JsonSerializer.Serialize(this, new JsonSerializerOptions { WriteIndented = true });
    }
}
```

```python
from langchain_ollama import OllamaEmbeddings
from langchain_community.vectorstores import Chroma

# Setting up langchain vectore store....Should make for every AI Agent using framework?
embeddings = OllamaEmbeddings(model="mxbai-embed-large", keep_alive=7200)

def Build_Collection(collection_name, embedding_function):
    collection = Chroma(
        collection_name=collection_name,
        embedding_function=embeddings
    )

    return collection
```

```python

# Builds a collection from a JSON
def Build_Profile_Collection(behavior_profile):

    # Extracting Json profile
    collection_name = "scenario_" + behavior_profile["Name"].lower().replace(" ", "_")
    scenario_chunks = behavior_profile["ScenarioChunks"]

    # Creating Collection for Behavior Profile
    collection = Build_Collection(collection_name, embeddings)

    # Creating Unique IDs for each scenario chunk
    ids = []
    for i in range(len(scenario_chunks)):
        ids.append(str(i))

    # Creating metadata for each scenario chunk
    metadatas = []
    for i in range(len(scenario_chunks)):
        metadatas.append({
            "id": str(i),
            "profile_name": behavior_profile["Name"]
        })

    # Adding chunks to database
    collection.add_texts(
        texts=scenario_chunks,
        ids=ids,
        metadatas=metadatas
    )

    return collection
```

```python
import textwrap
import time


def extract_game_context_from(scenario):
    game_context = ""

    # Agent Context
    agent = scenario["Agent"]

    game_context += "AGENT CHARCTER:\n"

    for context in agent["Context"]:
        game_context += context + "\n"

    game_context += "\n"

    # Relevant Character Contexts
    game_context += "RELEVENT CHARACTERS: \n"
    for character in scenario["Characters"]:
        game_context += "Character: " + character["Name"] + "\n"

        for context in character["Context"]:
            game_context += context + "\n"

        game_context += "\n"

    return game_context



def extract_valid_states_from(scenario):
    valid_states_context = ""

    for state in scenario["ValidTransitionStates"]:
        state_name = state["Name"]
        state_description = state["Description"]

        valid_states_context += "State: " + state_name + "\n" + "Description: " + state_description + "\n\n"

    return valid_states_context



def select_state(scenario, behavior_profile, collection, model, temperature=0.1, debug=False):

    # Converting to single string
    game_context = extract_game_context_from(scenario)
    valid_states_context = extract_valid_states_from(scenario)


    # Retrieval
    retrieved_chunks = collection.similarity_search_with_relevance_scores(
        game_context,
        k=3,
        score_threshold=0.4
    )

    retrieved_scenario_chunks = ""

    for document, score in retrieved_chunks:
        retrieved_scenario_chunks += document.page_content + "\n"

        if debug:
            print("ID:", document.metadata["id"])
            print("Chunk:", document.page_content)
            print("Score:", score)

    prompt= ""

    # run without scenario chunks
    if LLM_ONLY:
        # Prompt
        prompt = (
        "Behavior profile of the Agent:\n"
        f"{behavior_profile["Description"]}\n\n"
        "Current runtime game context:\n"
        f"{game_context}\n\n"
        "With the above context select the best state for the agent to transition to.\n"
        "Make the most suitable decision from the perspective of the agent\n"
        "Only select a state from the below valid states.\n"
        "Select only ONE of the valid states.\n"
        "Do not select state/action that is impossible for the agent to perform.\n"
        "Reponse should be limited to only the selected state and the target, nothing else.\n"
        "If there are no valid states repond with INVALID\n\n"
        "Valid States:\n"
        f"{valid_states_context}"
        "\n"
        "Follow this response format:\n"
        "STATE: <state>\n"
        "TARGET: <target>\n"
        "If the selected state does not require a target respond with this:\n"
        "STATE: <state>\n"
        "TARGET: NONE\n"
        "If there are no valid states repond with:\n"
        "STATE: INVALID\n"
        "TARGET: NONE\n"
        )

    # run query normally
    else:
        # Prompt
        prompt = (
        "Behavior profile of the Agent:\n"
        f"{behavior_profile["Description"]}\n\n"
        "Current runtime game context:\n"
        f"{game_context}\n\n"
        "Scenario based behavioral context:\n"
        f"{retrieved_scenario_chunks}\n"
        "With the above context select the best state for the agent to transition to.\n"
        "Make the most suitable decision from the perspective of the agent\n"
        "Only select a state from the below valid states.\n"
        "Select only ONE of the valid states.\n"
        "Do not select state/action that is impossible for the agent to perform.\n"
        "Reponse should be limited to only the selected state and the target, nothing else.\n"
        "If there are no valid states repond with INVALID\n\n"
        "Valid States:\n"
        f"{valid_states_context}"
        "\n"
        "Follow this response format:\n"
        "STATE: <state>\n"
        "TARGET: <target>\n"
        "If the selected state does not require a target respond with this:\n"
        "STATE: <state>\n"
        "TARGET: NONE\n"
        "If there are no valid states repond with:\n"
        "STATE: INVALID\n"
        "TARGET: NONE\n"
        )

    if debug:
        print("_______________\n")
        print("Prompt:")
        print(prompt, "\n")

    # LLM
    start_time = time.perf_counter()

    response = alpaca.chat(
        model=model,
        messages=[ {"role": "user", "content": prompt}],
        options={ "temperature": temperature},
    )

    response_time = time.perf_counter() - start_time

    return response, response_time
```

#Result Printing

```python
import csv

testResults = []

# Builds CSV Result file giving all statistics of every recorded query.
def Output_TestResultCSV():
    if LLM_ONLY is True:
        filename = "TestResults_LLM_ONLY.csv"
    else:
        filename = "TestResults.csv"

    # Build CSV
    with open(filename, mode="w", newline="", encoding="utf-8") as file:
        writer = csv.DictWriter(file, fieldnames=testResults[0].keys())

        writer.writeheader()
        writer.writerows(testResults)

    print("Results saved to ", filename)

# Builds a CSV Result file showing stats to compare the models.
def Output_ModelComparisonStats():
    if LLM_ONLY is True:
        filename = "ModelComparisons_LLM_ONLY.csv"
    else:
        filename = "ModelComparisons.csv"

    model_stats = {}
    model_stats_output = []

    # Collecting stats
    for result in testResults:
        if result["Model"] not in model_stats:
            model_stats[result["Model"]] = {
                "Model": result["Model"],
                "Behavior": result["Behavior"],
                "Recorded Count": 0,
                "Total Duration Sum": 0,
                "Load Duration Sum": 0,
                "Prompt Processing Sum": 0,
                "Token Generation Sum": 0,
                "Prompt Tokens Sum": 0,
                "Generated Tokens Sum": 0
            }

        # Collecting sum of each stat
        model_stats[result["Model"]]["Recorded Count"] += 1
        model_stats[result["Model"]]["Total Duration Sum"] += result["Total Duration"]
        model_stats[result["Model"]]["Load Duration Sum"] += result["Load Duration"]
        model_stats[result["Model"]]["Prompt Processing Sum"] += result["Prompt Processing"]
        model_stats[result["Model"]]["Token Generation Sum"] += result["Token Generation"]
        model_stats[result["Model"]]["Prompt Tokens Sum"] += result["Prompt Tokens"]
        model_stats[result["Model"]]["Generated Tokens Sum"] += result["Generated Tokens"]

    for stat in model_stats.values():
        average_stats = {
            "Model": stat["Model"],
            "Ave Total Duration": stat["Total Duration Sum"] / stat["Recorded Count"],
            "Ave Load Duration": stat["Load Duration Sum"] / stat["Recorded Count"],
            "Ave Prompt Processing": stat["Prompt Processing Sum"] / stat["Recorded Count"],
            "Ave Token Generation": stat["Token Generation Sum"] / stat["Recorded Count"]
        }

        model_stats_output.append(average_stats)

    # Build CSV
    with open(filename, mode="w", newline="", encoding="utf-8") as file:
        writer = csv.DictWriter(file, fieldnames=model_stats_output[0].keys())

        writer.writeheader()
        writer.writerows(model_stats_output)

    print("Results saved to ", filename)
```

```python


def Query_Scenario_List(scenario_list, behavior_profile, collection, model):
    i = 1
    for scenario in scenario_list:
        response, response_time = select_state(
            scenario,
            behavior_profile,
            collection,
            model=model,
            temperature=0.1,
            debug=False
        )

        # Saving for CSV Print
        result = {
          "Model": model,
          "Behavior": behavior_profile["Name"],
          "Scenario": i,
          "Response": response["message"]["content"],
          "Response Time": response_time,
          "Total Duration": response["total_duration"] / 1e9,
          "Load Duration": response["load_duration"] / 1e9,
          "Prompt Processing": response["prompt_eval_duration"] / 1e9,
          "Token Generation": response["eval_duration"] / 1e9,
          "Prompt Tokens": response.get("prompt_eval_count", 0),
          "Generated Tokens": response.get("eval_count", 0)
        }
        testResults.append(result)

        # Console test print
        print("--------------------------")
        print("Behavior: ", behavior_profile["Name"])
        print("Model: ", model)
        print("Response to Scenario ", i, ":")
        print(response["message"]["content"], "\n")
        print(f"Response Time: {response_time:.3f} seconds")
        print(f"Total Ollama Duration: {response['total_duration'] / 1e9:.3f}s")
        print(f"Model Load Duration: {response['load_duration'] / 1e9:.3f}s")
        print(f"Prompt Processing: {response['prompt_eval_duration'] / 1e9:.3f}s")
        print(f"Token Generation: {response['eval_duration'] / 1e9:.3f}s\n")
        print(f"Generated Tokens: {response["eval_count"]}")

        i = i + 1
```
